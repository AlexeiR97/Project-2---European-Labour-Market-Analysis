```python
import json
```

```python
input_file = "../PowerBI/geo_files/NUTS_RG_20M_2021_4326_LEVL_2.geojson"
```

```python
with open(input_file, 'r', encoding='utf-8') as f:
    data = json.load(f)

def filter_and_save(country_code, output_filename):
    filtered_features = []
    for feature in data["features"]:
        props = feature.get("properties", {})
        if props.get("CNTR_CODE") == country_code:
            filtered_features.append(feature)

    filtered_data = {
        "type": "FeatureCollection",
        "features": filtered_features
    }
    
    with open(output_filename, 'w', encoding='utf-8') as f:
        json.dump(filtered_data, f, ensure_ascii=False, indent=2)
    
    print(f"{country_code}: {len(filtered_features)} features saved to {output_filename}")
```

```python
filter_and_save("ES", "../PowerBI/geo_files/spain_nuts2.geojson")
filter_and_save("IT", "../PowerBI/geo_files/italy_nuts2.geojson")

print("Filtering and saving completed.")
```
