So, many columns are duplicates or simply not encoded. I would like to remove Age Class as its codified column "age" is easy to understand already, and also I will do modification to age column (Y_GE - I just don`t like how it sounds, lets rename it to Y_O (years_over)).

VERY IMPORTANT AND EASILY MISSED THING - AGGREGATED CATEGORIES!!!! We don`t need all these Y30-49, Y50-64, Y65-84, Y_LT15,  we already have separate categories for all of this, so we completely remove these completely.

Y_GE100 and Y_LT5 are aggregated, but their numbers are both under "Not Applicable" categories, so we can leave these categories. (100+ old people and babies don`t work, so no work category)

```python
aggregated_ages = ['Y15-29', 'Y30-49', 'Y50-64', 'Y65-84', 'Y_GE85', 'Y_LT15']
df2 = df2[~df2['age'].isin(aggregated_ages)]
print(df2['age'].unique())
```

As we can ses above - I renamed the age categories to make them more consistent and easier to understand. Now, 'Y_GE100' is renamed to 'Y_O100' (indicating age 100 or older) and 'Y_GE85' is renamed to 'Y_O85' (indicating age 85 or older). 
This makes the age categories more intuitive and easier to interpret and I can remove Age class completely

Lets create a dictionary of columns to drop

Lets write a quick cicle to remove TOTAL rows from our Data.

```python
cols_to_clean = ['wstatus', 'Labour force and employment status', 'isco08', 
                 'International Standard Classification of Occupations 2008 (ISCO-08)',
                 'Labour force and employment status',
                 'age', 'sex', 'geo']

for col in cols_to_clean:
    df2 = df2[df2[col].str.lower() != 'total']
```

and a small line to remove rows with "T" (meaning Total)

```python
df2 = df2[df2['sex'].str.strip().str.upper() != 'T']
```

```python
df2 = df2[(df2['OBS_VALUE'].notna()) & (df2['OBS_VALUE'] != 0)]
```
